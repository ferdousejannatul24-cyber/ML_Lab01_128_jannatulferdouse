```python
#12. Find the minimum mark and standard deviation.
#13. Add 5 bonus marks to every student without changing the original array.
#14. Count how many students passed.
#15. Create a 2 × 3 matrix from the marks.
import numpy as np

np.random.seed(42)
marks = np.array([55, 72, 48, 90, 66, 81])

print('Marks:', marks)
print('Mean:', marks.mean())
print('Highest:', marks.max())
print('Passed:', marks[marks >= 50])

print('Minimum:', marks.min())
print('Standard Deviation:', marks.std())

bonus_marks = marks + 5
print('Marks with Bonus:', bonus_marks)
print('Original Marks:', marks)

passed_count = np.sum(marks >= 50)
print('Number of Passed Students:', passed_count)

matrix = marks.reshape(2, 3)
print('2 × 3 Matrix:')
print(matrix)
```

```python
#16. Which columns are features? Which column is the label?
#17. How many samples and columns are present?
#18. Filter students with attendance of at least 80.
#19. Calculate average study hours for students who passed.
#20. Sort the DataFrame by attendance in descending order.
#21. Does this tiny dataset prove that snack purchases cause failure? Explain why or why not.
import pandas as pd

data = {
    'student': ['Asha', 'Babu', 'Chayan', 'Dipa', 'Emon', 'Farah'],
    'study_hours': [2, 5, 1, 7, 3, 6],
    'sleep_hours': [8, 6, 9, 5, 7, 6],
    'attendance': [70, 88, 55, 95, 75, 91],
    'snacks': [3, 2, 5, 1, 4, 2],
    'passed': ['Yes', 'Yes', 'No', 'Yes', 'No', 'Yes']
}

df = pd.DataFrame(data)
display(df)
print(df.info())
print(df.describe())

print("Features:", ['study_hours', 'sleep_hours', 'attendance', 'snacks'])
print("Label:", 'passed')

print("Samples:", df.shape[0])
print("Columns:", df.shape[1])

print(df[df['attendance'] >= 80])

print("Average study hours for passed students:", df[df['passed'] == 'Yes']['study_hours'].mean())

print(df.sort_values(by='attendance', ascending=False))

print("No, this tiny dataset does not prove that snack purchases cause failure.")
```

```python
#Activity D — Reproducibility Experiment
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
    return np.random.randint(40, 101, size=5)

print('Seeded run 1:', generate_scores(True))
print('Seeded run 2:', generate_scores(True))
print('Unseeded run 1:', generate_scores(False))
print('Unseeded run 2:', generate_scores(False))
```

```python
#Run the cell twice. Record which outputs remain identical. Then explain: if two groups claim different
#accuracies because they used different random splits, can we fairly compare their models?
import random
import numpy as np

def generate_scores(use_seed):
    if use_seed:
        random.seed(42)
        np.random.seed(42)
        return np.random.randint(40, 101, size=5)
    return np.random.randint(40, 101, size=5)

seeded_run_1 = generate_scores(True)
seeded_run_2 = generate_scores(True)
unseeded_run_1 = generate_scores(False)
unseeded_run_2 = generate_scores(False)

print('Seeded run 1:', seeded_run_1)
print('Seeded run 2:', seeded_run_2)
print('Unseeded run 1:', unseeded_run_1)
print('Unseeded run 2:', unseeded_run_2)

print('Seeded outputs identical:', np.array_equal(seeded_run_1, seeded_run_2))
print('Unseeded outputs identical:', np.array_equal(unseeded_run_1, unseeded_run_2))

print('Models using different random splits cannot be fairly compared directly.')
print('They should use the same random split or the same fixed random seed.')
```
